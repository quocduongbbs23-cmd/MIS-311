## 3. Descriptive Statistics and Insight Generation

After confirming that the dataset is clean and ready for analysis, descriptive statistics were used to summarise the main characteristics of the data. The analysis focuses on identifying meaningful patterns related to student placement outcomes and supporting these findings with appropriate visualisations.

### 3.1 Distribution of Placement Outcomes

```python
# Count the number of placed and not placed students
placement_counts = df['placed'].value_counts().sort_index()

print("Number of students by placement status:")
print(f"Not Placed (0): {placement_counts[0]} students")
print(f"Placed (1): {placement_counts[1]} students")
```

```python
# Calculate the percentage of students by placement status
placement_percentages = df['placed'].value_counts(normalize=True).sort_index() * 100

print("Percentage of students by placement status:")
print(f"Not Placed (0): {placement_percentages[0]:.2f}%")
print(f"Placed (1): {placement_percentages[1]:.2f}%")
```

The placement outcome is relatively balanced across the dataset. Out of **2,000 students**, **1,023 students (51.15%)** were not placed, while **977 students (48.85%)** were placed.

This near-balanced distribution is useful for further comparison because both placement groups are sufficiently represented in the dataset, allowing meaningful descriptive analysis of differences between placed and non-placed students.

```python
# Visualise the number of students by placement status
placement_labels = ['Not Placed', 'Placed']

plt.figure(figsize=(7, 5))
plt.bar(placement_labels, placement_counts.values)

plt.title('Distribution of Student Placement Outcomes')
plt.xlabel('Placement Status')
plt.ylabel('Number of Students')

for index, value in enumerate(placement_counts.values):
    plt.text(index, value + 10, str(value), ha='center')

plt.show()
```

### 3.2 Summary Statistics of Key Variables

```python
# Select key numerical variables for descriptive statistics
selected_columns = [
    'cgpa',
    'attendance',
    'internships',
    'projects',
    'github_projects',
    'python_skill',
    'java_skill',
    'ml_skill',
    'web_dev_skill',
    'leadership',
    'teamwork',
    'problem_solving'
]

# Generate descriptive statistics for selected variables
summary_stats = df[selected_columns].describe().T

# Keep the most useful statistics and round values to 2 decimal places
summary_stats = summary_stats[['mean', 'std', 'min', '25%', '50%', '75%', 'max']].round(2)

summary_stats
```

The summary statistics provide an overall picture of students’ academic performance, practical experience, and skill levels. On average, students had a **CGPA of 7.48** and an **attendance rate of 79.76%**, suggesting a generally moderate-to-good academic profile across the dataset.

In terms of employability-related experience, students completed an average of **1.47 internships**, **3.09 projects**, and **4.01 GitHub projects**. The average scores for technical and soft-skill indicators were mostly between **4.33 and 4.53**, showing that these capabilities vary across students and may be useful for deeper comparison by placement status.

### 3.3 Comparison of Academic and Soft-Skill Indicators by Placement Status

```python
# Compare key academic and soft-skill indicators by placement status
academic_softskill_columns = [
    'cgpa',
    'attendance',
    'communication_skill',
    'teamwork',
    'problem_solving'
]

academic_softskill_comparison = (
    df.groupby('placed')[academic_softskill_columns]
    .mean()
    .round(2)
)

academic_softskill_comparison.index = ['Not Placed', 'Placed']

academic_softskill_comparison
```

The comparison shows that placed students had a higher average **CGPA (7.62)** than non-placed students **(7.35)**. They also recorded stronger average scores in **teamwork (4.57 vs. 4.11)** and **problem-solving (4.67 vs. 4.29)**.

In contrast, the differences in **attendance** and **communication skill** were very small, suggesting that these two indicators do not show a clear descriptive gap between the two placement groups in this dataset.

```python
# Visualise the average CGPA by placement status
cgpa_by_placement = df.groupby('placed')['cgpa'].mean().round(2)

cgpa_labels = ['Not Placed', 'Placed']

plt.figure(figsize=(7, 5))
plt.bar(cgpa_labels, cgpa_by_placement.values)

plt.title('Average CGPA by Placement Status')
plt.xlabel('Placement Status')
plt.ylabel('Average CGPA')

for index, value in enumerate(cgpa_by_placement.values):
    plt.text(index, value + 0.03, str(value), ha='center')

plt.ylim(0, 9)
plt.show()
```

```python
# Visualise average teamwork and problem-solving scores by placement status
soft_skill_comparison = academic_softskill_comparison[
    ['teamwork', 'problem_solving']
]

ax = soft_skill_comparison.plot(kind='bar', figsize=(8, 5))

plt.title('Average Teamwork and Problem-Solving Scores by Placement Status')
plt.xlabel('Placement Status')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.ylim(0, 6)
plt.legend(title='Soft-Skill Indicators')

# Add value labels above each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)

plt.show()
```

### Key Insight 1: Academic Performance and Selected Soft Skills Differ by Placement Status

Placed students demonstrated stronger average results in several important indicators. Their average **CGPA was 7.62**, compared with **7.35** among non-placed students, while their average **teamwork** and **problem-solving** scores were also higher at **4.57** and **4.67**, respectively. This suggests that placement outcomes in this dataset may be associated with a combination of academic performance and selected soft skills, rather than academic results alone.

### 3.4 Placement Rate by Internship Experience

```python
# Create internship experience groups
df['internship_group'] = pd.cut(
    df['internships'],
    bins=[-1, 0, 1, df['internships'].max()],
    labels=['0 Internships', '1 Internship', '2+ Internships']
)

# Check the number of students in each internship group
print("Number of students in each internship group:")
print(df['internship_group'].value_counts().sort_index())
```

```python
# Calculate placement rate for each internship experience group
internship_placement_rate = (
    df.groupby('internship_group', observed=True)['placed']
    .mean()
    .mul(100)
    .round(2)
)

print("Placement rate by internship experience group:")
print(internship_placement_rate)
```

```python
# Visualise placement rate by internship experience group
plt.figure(figsize=(8, 5))
plt.bar(internship_placement_rate.index, internship_placement_rate.values)

plt.title('Placement Rate by Internship Experience')
plt.xlabel('Internship Experience')
plt.ylabel('Placement Rate (%)')
plt.ylim(0, 65)

# Add percentage labels above each bar
for index, value in enumerate(internship_placement_rate.values):
    plt.text(index, value + 1, f'{value:.2f}%', ha='center')

plt.show()
```

The placement rate increased across the three internship experience groups. Students with **0 internships** had a placement rate of **41.08%**, which rose to **46.96%** for students with **1 internship** and reached **54.53%** for those with **2 or more internships**.

### Key Insight 2: Greater Internship Experience Is Associated with Higher Placement Rates

Students with more internship experience showed higher placement rates in this dataset. The placement rate increased from **41.08%** among students with **0 internships** to **54.53%** among students with **2 or more internships**. This suggests that practical work experience may be an important employability-related factor associated with stronger placement outcomes.
